```python
import numpy as np
```

NEURAL NERTWOK

```python
#inputs
x=np.array([3,5,7,9])

#true label
y=np.array([12])

#learning rate
learning_rate=0.01

#first layer weight and bias
layer1_w=np.array([[1,2,3,4],
                   [5,6,7,8]])
layer1_b=np.array([1,2])

#second layer weight and bias
layer2_w=np.array([[2,4],
                   [3,5]])
layer2_b=np.array([1,3])

#output neuron weight and bias
output_w=np.array([5,6])
output_b=np.array([2])

#TRAINING LOOP

for epoch in range(1000):

  #first pass
  z1=np.dot(x,layer1_w.T)+layer1_b
  a1=np.maximum(0,z1) #activation function
  relu_gradient1=(z1>0).astype(float)  #relu gradient

  #second pass
  z2=np.dot(a1,layer2_w)+layer2_b
  a2=np.maximum(0,z2)  #activation function
  relu_gradient2=(z2>0).astype(float) #relu gradient

  #output neuron pass

  z3=np.dot(a2,output_w.T)+output_b
  a3=z3 #linear activation

  #loss
  loss=0.5*(y-a3)**2

  #error

  #output neuron error
  delta_output=a3-y
  #error of the neurons in the second hidden layer
  delta_layer2=delta_output*output_w*relu_gradient2

  #error of the neurons in the first hiddn layer
  delta_layer1=np.dot(delta_layer2,layer2_w)*relu_gradient1

  #bias gradient
  output_b_grad=delta_output
  layer2_b_grad=delta_layer2
  layer1_b_grad=delta_layer1

  #weight gradient
  output_w_grad=delta_output*a2 #weight gradient of the output neuron

  layer2_w_grad=np.outer(delta_layer2,a1) #weight gradient of the neurons in the second hidden layer

  layer1_w_grad=np.outer(delta_layer1,x)  #weight gradient of the neurons in the first hidden layer

  #Update

  output_w=output_w-learning_rate*output_w_grad  #output neuron weight update
  output_b=output_b-learning_rate*output_b_grad  #output neuron bias update

  layer2_w=layer2_w-learning_rate*layer2_w_grad  #hidden layer2 weight update
  layer2_b=layer2_b-learning_rate*layer2_b_grad  #hidden layer2 bias update

  layer1_w=layer1_w-learning_rate*layer1_w_grad  #hidden layer1 weight update
  layer1_b=layer1_b-learning_rate*layer1_b_grad  #hidden layer1 bias update

  if epoch %300==0:
    print(f'epoch:{epoch}, loss:{loss.item():.2f}')




```

```python
np.savez('neural_nertwork_model10.npz',
         weight_hidden1=layer1_w,
         bias_hidden1=layer1_b,
         weight_hidden2=layer2_w,
         bias_hidden2=layer2_b,
         output_w_new=output_w,
         output_b_new=output_b)
```
